```python
import os, glob, shutil, zipfile
R = "/content/Retrieval-based-Voice-Conversion-WebUI"
os.makedirs(f"{R}/assets/hubert_base", exist_ok=True)
os.makedirs(f"{R}/assets/rmvpe", exist_ok=True)

for name in ["hubert_base", "rmvpe"]:
    tmp = f"/content/_{name}_tmp"
    shutil.rmtree(tmp, ignore_errors=True)
    with zipfile.ZipFile(f"/content/drive/MyDrive/{name}.zip") as z:
        z.extractall(tmp)
    # 파일이 들어 있는 가장 안쪽 폴더를 찾아 assets로 옮김
    root = tmp
    while True:
        items = os.listdir(root)
        if len(items) == 1 and os.path.isdir(f"{root}/{items[0]}"):
            root = f"{root}/{items[0]}"
        else:
            break
    for f in os.listdir(root):
        shutil.move(f"{root}/{f}", f"{R}/assets/{name}/{f}")

!ls -la {R}/assets/hubert_base
!ls -la {R}/assets/rmvpe
```
